```python
df_ml = df.dropna().copy()

print("Original dataset rows:", len(df))
print("Machine Learning dataset rows:", len(df_ml))

df_ml.head()
```

```python
print("Dataset Shape:", df_ml.shape)

print("\nColumns:")
print(df_ml.columns.tolist())

print("\nMissing Values:")
print(df_ml.isnull().sum())

df_ml.tail()
```

```python
features = [
    'Open',
    'High',
    'Low',
    'Close',
    'Volume',
    'MA_7',
    'MA_30',
    'MA_50',
    'Daily_Return',
    'Volatility'
]

X = df_ml[features]

X.head()
```

```python
y = df_ml['Target']

y.head()
```

```python
split = int(len(df_ml) * 0.8)

X_train = X.iloc[:split]
X_test = X.iloc[split:]

y_train = y.iloc[:split]
y_test = y.iloc[split:]

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

```python
print("Training period:")
print(df_ml['Date'].iloc[0], "to", df_ml['Date'].iloc[split - 1])

print("\nTesting period:")
print(df_ml['Date'].iloc[split], "to", df_ml['Date'].iloc[-1])
```

```python
print("Missing values in X_train:", X_train.isnull().sum().sum())
print("Missing values in X_test:", X_test.isnull().sum().sum())
print("Missing values in y_train:", y_train.isnull().sum())
print("Missing values in y_test:", y_test.isnull().sum())
```

```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
```

```python
model_lr = LinearRegression()

model_lr.fit(X_train, y_train)

pred_lr = model_lr.predict(X_test)

print("Linear Regression model trained successfully!")
print("Total predictions:", len(pred_lr))
```

```python
model_rf = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model_rf.fit(X_train, y_train)

pred_rf = model_rf.predict(X_test)

print("Random Forest model trained successfully!")
print("Total predictions:", len(pred_rf))
```

```python
model_gb = GradientBoostingRegressor(
    n_estimators=100,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

model_gb.fit(X_train, y_train)

pred_gb = model_gb.predict(X_test)

print("Gradient Boosting model trained successfully!")
print("Total predictions:", len(pred_gb))
```

```python
comparison = pd.DataFrame({
    'Actual Price': y_test.values,
    'Linear Regression': pred_lr,
    'Random Forest': pred_rf,
    'Gradient Boosting': pred_gb
})

comparison.head(10)
```

```python
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))

print("Linear Regression predictions:", len(pred_lr))
print("Random Forest predictions:", len(pred_rf))
print("Gradient Boosting predictions:", len(pred_gb))
```

```python
import numpy as np
import pandas as pd

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

```python
def evaluate_model(name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)

    rmse = np.sqrt(mean_squared_error(y_true, y_pred))

    r2 = r2_score(y_true, y_pred)

    return {
        'Model': name,
        'MAE': mae,
        'RMSE': rmse,
        'R2 Score': r2
    }
```

```python
results = []

results.append(
    evaluate_model('Linear Regression', y_test, pred_lr)
)

results.append(
    evaluate_model('Random Forest', y_test, pred_rf)
)

results.append(
    evaluate_model('Gradient Boosting', y_test, pred_gb)
)

results_df = pd.DataFrame(results)

results_df
```

```python
best_model_row = results_df.loc[results_df['MAE'].idxmin()]

print("Best Model based on MAE:", best_model_row['Model'])
print("MAE:", round(best_model_row['MAE'], 4))
print("RMSE:", round(best_model_row['RMSE'], 4))
print("R2 Score:", round(best_model_row['R2 Score'], 4))
```
